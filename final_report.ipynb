The main aspect of the winter weather that causes trouble for rail transit is the freezing temperatures. When temperatures get as low as they did on January 6, it can cause ice to form on the tracks and catenary wires above the tracks. Ice on the tracks creates issues for the rail switches as they can literally freeze in place. Areas with colder climates that experience these issues more often have ways to mitigate these effects, mainly through heating the rails. Additionally, ice on the catenary wires can interfere with electricity transfer between the train and the wires.
These issues notably affected Utrecht Centraal (How do we know this?). Utrecht Centraal is the largest rail station in The Netherlands by passenger throughput [SOURCE] and is located right where the coldest temperatures were during the storm. Additionally, given the high passenger numbers, there is a large amount of rail infrastructure at the station, including around 70 rail switches, which are vulnerable to the cold temperatures.

The three indicators and justification used are as follows:
- Delay indicator: 
    - Every arrival- and departure delay is flagged if the delay column > 0. 
    - An additional binary indicator is more important, however, which is the arrival and departure delay being bigger or equal to three (3) minutes or 180 seconds. This is the delay value indicated in NS' performance agreements as communicated to passengers and stakeholders [3].
- Number of cancellations:
    - A train is cancelled if either arrival or departure is cancelled.
    - Cancellations can be per hour, day, week of day or the average over the analysis period.
- Punctuality performance:
    - Punctuality of a train is the number of on-time events (on-time indicating delays being smaller than 180 seconds) divided by the total scheduled events.
    - Punctuality can differ at arrival, departure and a total punctuality figure.
    - Keeping non-cancelled trains in this dataset

The delay indicator entails how much and how severe a disruption is, where the number of cancellations complement this. Punctuality answers the question how often the services meet the performance standard. The delay measures the magnitude and distribution of operational disruption, whereas the punctuality measures the proportion of services which meet a predefined performance standard.

### 3.2.4 Delay absorption and propagation

The first plot (Fig. 3.2.9) shows delay recovery or delay increase in per weekday and per hour. The left plots show all delayed trains using a net-delay-change and the right plots show delay absorption of late-arriving trains only.
- In the left panels, blue bars indicate a net delay reduction at the stop and the red bars a net delay increase.
- In the right panels, the blue indicates trains whose delay is absorbed and red trains whose delay propagated.

The net delay change equals the difference between the departure and arrival delay; if net delay change is bigger than zero delays are absorbed or dissolved:
- departure delay - departure delay is the actual arrival delay removed.

The delay absorption equals the difference between the departure and arrival delay, divided by the departure delay. 
- (arrival delay - departure delays) / arrival delays gives the share of arrival delay removed

On average, the net delay change is slightly negative (-0.08 min/train), so stops increase delay by a very small, immeasurable, margin. The value per train is small because most trains have no delay to recover. Measured in delay minutes, 12.2% of arrival delay is absorbed, rising to 31.0% for late-arriving trains. From this point on absorption is only calculated for late-arriving trains, as trains arriving on time have no delay to absorb and dilute the statistic. The net delay change is calcualted for all trains, as it also captures delay which is generated by the station.

The plots first column on the statistics of the net delay reduction is an indicative plot showing the minimalistic influence of net delay change on the total dataset; the delay averages down to zero in a sufficiently large dataset.

Absorption is an interesting statistic to show in this stage. Absorption of late trains is highest on Wednesdays (around 36%) and lowest on Fridays (around 27%). By hour, absorption is highest is highest in the early morning and decreases to around 25% in the afternoon. The recovery of capacity is higher in the morning than in the afternoon peak.

### 3.2.5 Correlation and dwell-time heatmaps

This section looks at two things; how strongly the weekly operational metrics are related and to what extent the scheduled dwell time at Utrecht Centraal determines or explains the absorption.

Dwell time determines the available buffer for trains to keep on time. 

Three heatmaps are used: a weekly correlation matrix, the delay absorption of late-arriving trains by dwell time and departure hour and the mean net delay change of all trains by dwell time and departure hour.

```python
analyse_correlations(train)
```

##### 1) Weekly correlation (Fig. 3.2.11)

Weekly arrival and departure delays are almost perfectly correlated (r = 0.92), which is almost trivial. Weeks with many late arrivals also have many late departures.

The cancellation rate has a moderate positive correlation with delays, so poor weeks tend to be poor in arrival and departure delays. The number of scheduled trains is practically unrelated to delays and weakly related to cancellations (r = -0.16). Disruptive weeks probably drive both delays and cancellations.

##### 2) Delay absorption heatmap (Fig. 3.2.12)

Absorption among late-arriving trains depends mainly on the scheduled dwell time. For a dwell time of 0-1 minute, absorption is negative in all hours with sufficient data (-14% to -33% increase in delay), so these trains leave with more delay than they arrive with. At a scheduled dwell time of 2 minutes the absorption nears zero and steadily rises afterwards from 18% at 3 minutes to 72% at 8-9 minutes.

The departure hour has a second effect. For the same dwell time, the absorption is the highest in the early morning and lowest in the early afternoon; except for scheduled dwell time between 0-1 minutes. The higher afternoon absorption and lower peak-hour absorption (propagation) reflect higher loads on the station and trains in the peak-hours.

##### 3) Net delay change heatmap (Fig. 3.2.13)

The net delay change show the same gradient as in 2) but in minutes per train. For 0-1 minutes of dwell time, net delay change is positive in all hours (+0.8 to +1.8 min), so delay is generated at these stops. At 2-4 minutes the net change is close to zero and from 4 minutes onwards the net delay change is negative.

The delays turn positive in the afternoon, as in line with the low absorption in the previous heatmap. For long dwell times in the late evening, the net delay is positive; these cells may reflect turnaround or other out of time events and must be treated with caution. 

The small average net change (-0.08 min/train) in Section 3.2.4 clearly hides two opposing groups; trains with a short scheduled dwell time which add delay and trains with a longer dwell time which absorb delay.

The following conclusions can be drawn:
1) Delay absorption is mainly determined by the buffer in the timetable. Below 3 minutes of scheduled dwell time, the stop produces delay. Above 5 minutes it removes the largest share of delays.
2) At equal dwell times, the absorption is lower in the afternoon peak. A loaded network thus reduces the effectiveness of the buffer.
3) Weekly delays and cancellations move together.
4) Differences between weekdays and hours in absorption and net change are both dependent on the buffer and network load, not solely on the level of incoming delay.

### 3.4.6 Conclusions

Over the baseline study period, delays are short and mostly small. The mean arrival delay is within the minute per train, depending on the hour. More than 90% of the trains arrive and depart within 3 minutes of schedule. Of the approximately 9,000 trains with a delay at either arrival or departure, 30% are delayed only at arrival. Cancellations vary strongly throughout the day, but most occur in the afternoon.

**A typical day** therefore starts with high punctuality and few delays. Delays gradually build up through the day, peak in the afternoon rush period (16:00 - 19:00) and increase again in the late evening. Cancellations peak earlier than delays, already in the early afternoon.

Across the week, Tuesday is the least punctual day and Sunday the most punctual. Across the period, delays are lowest around the turn of the year and highest in mid-February, after which they stabilize. Under normal conditions a delay is small, recurring and has a predictable temporal pattern. Large deviations from this pattern coincide with disruptions and are accompanied by increased cancellations.

The stop of Utrecht Centraal is a recovery point, but only to a limited extent. For late arriving trains, aprroximately one-third of the arrival delay can be removed. This recovery mainly occurs when the timetable provides sufficient scheduled dwell time, particularly at least three minutes, while recovery is more limited during the busy afternoon peak. Trains with a scheduled dwell time of 0-2 minutes systematically accumulate additional delay.


However, differences in delay abosorption and net delay change between hours and weekdays should not be attributed solely to the level of incoming delay. THey are influenced by the interaction between the available buffer time and the operational network load. Periods with greater network pressure, such as the afternoon peak, may therefore provide fewer opportunities to use available dwell time for recovery. The most direct operational lever for improving punctuality at Utrecht Centraal is consequently the provision of sufficient timetable buffer, particularly during periods of high network load.

```python
# 1. Load the dataset

df = pd.read_csv('./data/weather_selected.csv')

# Clean column names by removing '#' and extra whitespace
df.columns = df.columns.str.replace('#', '').str.strip()

# 2. Process Dates and Hours
# KNMI hourly data uses hours 1 to 24. We convert hour 24 properly by adjusting HH - 1.
df['HH_adjusted'] = df['HH'] - 1
df['datetime'] = pd.to_datetime(df['YYYYMMDD'].astype(str), format='%Y%m%d') + pd.to_timedelta(df['HH_adjusted'], unit='h')

# Convert KNMI scaled values to standard units:
# T: Temperature in 0.1 °C -> divide by 10
# TD: Dew point temperature in 0.1 °C -> divide by 10
df['T_C'] = df['T'] / 10.0
df['TD_C'] = df['TD'] / 10.0

# RH: Hourly precipitation amount in 0.1 mm (-1 means <0.05 mm trace amount)
# We convert -1 to 0.0 for numerical plotting, and divide valid values by 10.
df['RH_mm'] = df['RH'].apply(lambda x: 0.0 if x == -1 else x / 10.0)

# 3. Filter data for the specified period
# Note: If testing with the sample snippet from Jan 2021, change dates to '2021-01-01' and '2021-01-14'.
start_date = '2026-01-01'
end_date = '2026-01-14'

mask = (df['datetime'] >= start_date) & (df['datetime'] <= end_date + ' 23:59:59')
df_filtered = df.loc[mask].copy()

# Select columns related to weather variables
columns_of_interest = ['datetime', 'T_C', 'TD_C', 'RH_mm', 'S', 'R', 'WW', 'VV', 'U']
df_weather = df_filtered[columns_of_interest].dropna(subset=['datetime'])
```

```python
def load_train_data(data='data/traindata_total.csv',
                    start_period='2025-12-15', end_period='2026-04-30'):
    """Load, clean and prepare the train dataset; does not create plots."""
    train = pd.read_csv(data)
    train['Service:Date'] = pd.to_datetime(train['Service:Date'])
    train = train[train['Service:Date'].between(start_period, end_period)].copy()

    train = train.drop(columns=['index', 'Service:RDT-ID']) # Drop columns not needed for analysis
    train = train.rename(columns={                          # Rename columns for clarity
        'Service:Type': 'Type', 'Service:Date': 'Date',
        'Stop:Arrival time': 'scheduled_arrival',
        'Stop:Departure time': 'scheduled_departure',
        'Stop:Arrival cancelled': 'arr_cancelled',
        'Stop:Departure cancelled': 'dep_cancelled',
        'Stop:Arrival delay': 'arr_delay',
        'Stop:Departure delay': 'dep_delay'
    })
    train.columns = (train.columns.str.replace('Service:', '', regex=False)
                     .str.replace('Stop:', '', regex=False)
                     .str.replace('RDT-ID', 'train_id', regex=False))

    drop_cols = [
        'Completely cancelled', 'Partly cancelled', 'Maximum delay',
        'Train number', 'Station name', 'Station code', 'Platform change',
        'Planned platform', 'Actual platform', 'Company'
    ]
    train = train.drop(columns=drop_cols)
    train = train[train['Type'].isin(['Intercity', 'Sprinter'])].copy()

    for col in ['scheduled_arrival', 'scheduled_departure']:
        train[col] = pd.to_datetime(train[col], utc=True).dt.strftime('%H:%M')
    
    # Creating actual departure and arrival times based on scheduled times and delays (actual = scheduled + delay)
    for label, delay_col in [('arrival', 'arr_delay'), ('departure', 'dep_delay')]:
        sched_col = f'scheduled_{label}'
        actual_col = f'actual_{label}'
        scheduled = pd.to_datetime(train[sched_col], format='%H:%M')
        actual = scheduled + pd.to_timedelta(train[delay_col], unit='m')
        train[sched_col] = scheduled.dt.strftime('%H:%M')
        train[actual_col] = actual.dt.strftime('%H:%M')

    # Creating arrival and departure hour columns for filtering and analysis
    train['arr_hour'] = pd.to_datetime(train['actual_arrival'], format='%H:%M', errors='coerce').dt.hour
    train['dep_hour'] = pd.to_datetime(train['actual_departure'], format='%H:%M', errors='coerce').dt.hour
    train = train[train['dep_hour'].between(5, 22) & train['arr_hour'].between(5, 22)].copy()
    train = train.sort_values(['Date', 'actual_arrival']).reset_index(drop=True)

    # Creating new variable columns as indicators to flag delay and cancellations
    train['arrival_delayed'] = train['arr_delay'] > 0
    train['departure_delayed'] = train['dep_delay'] > 0
    train['arrival_delayed_3'] = train['arr_delay'] >= 3
    train['departure_delayed_3'] = train['dep_delay'] >= 3
    train['cancelled'] = train['arr_cancelled'].fillna(False).astype(bool) | train['dep_cancelled'].fillna(False).astype(bool)
    train['week'] = train['Date'].dt.to_period('W').astype(str)
    return train

#Loading new dataframe

df_train = load_train_data(start_period= '2026-01-01', end_period= '2026-01-15')

#Combining the time (scheduled arrival) and date
df_train['Start_Datetime'] = pd.to_datetime(df_train['Date'].astype(str) + ' ' + df_train['scheduled_arrival'])
df_hour = df_train

#Round off hours to the nearest whole number
df_hour['Hour'] = df_hour['Start_Datetime'].dt.floor('h')

#Taking the total trains for each hour
trains = df_hour.groupby('Hour').count()

#Taking the total cancelled an delayed trains each hour
trains_cancelled = df_hour.groupby('Hour')['cancelled'].sum()
trains_arrival_delayed = df_hour.groupby('Hour')['arrival_delayed'].sum()
trains_departure_delayed = df_hour.groupby('Hour')['departure_delayed'].sum()

#Setting up the time period
time_period = pd.date_range(start='2026-01-01', end=pd.to_datetime('2026-01-15') + pd.Timedelta(hours=23), freq='h')

#Reindexing the trains to each hour
hourly_trains = trains['Type'].reindex(time_period, fill_value=0)
hourly_cancelled = trains_cancelled.reindex(time_period, fill_value=0)
hourly_arrival_delayed = trains_arrival_delayed.reindex(time_period, fill_value=0)
hourly_departure_delayed = trains_departure_delayed.reindex(time_period, fill_value=0)

#Making a function to convert to percentage
def total_to_percentage(input, total):
    return np.where(total == 0, 0, 100*input/total)

#Setting up hourly punctuality
hourly_punctuality = 0

```
